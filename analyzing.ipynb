```python
weather["timestamp"] = pd.to_datetime(
    weather["timestamp"],
    errors="coerce"
)

def convert_to_ist(row):
    if pd.isna(row["timestamp"]):
        return pd.NaT

    ts = row["timestamp"]

    if row["timestamp_tz"] == "UTC":
        return ts.tz_localize("UTC").tz_convert("Asia/Kolkata")

    elif row["timestamp_tz"] == "IST":
        return ts.tz_localize("Asia/Kolkata")

    else:
        return ts.tz_localize("Asia/Kolkata")

weather["timestamp_ist"] = weather.apply(convert_to_ist, axis=1)

weather["Date"] = pd.to_datetime(
    weather["timestamp_ist"],
    errors="coerce"
).dt.normalize()

weather_daily = (
    weather.groupby("Date", as_index=False)
    .agg(
        Avg_Temperature_C=("temperature_c", "mean"),
        Total_Rainfall_mm=("rainfall_mm", "sum"),
        Avg_Humidity=("humidity_percent", "mean"),
        Sensor_Count=("sensor_id", "nunique")
    )
)

print("Weather daily shape:", weather_daily.shape)
print("\nDuplicate dates:", weather_daily["Date"].duplicated().sum())

display(weather_daily.head())
```

```python
# Prepare dates for merging
price_mandi_daily["date"] = pd.to_datetime(
    price_mandi_daily["date"],
    errors="coerce"
).dt.normalize()

weather_daily["Date"] = pd.to_datetime(
    weather_daily["Date"],
    errors="coerce"
).dt.tz_localize(None)

# Merge weather data
final_data = price_mandi_daily.merge(
    weather_daily,
    left_on="date",
    right_on="Date",
    how="left"
)

# Remove duplicate date column
final_data = final_data.drop(columns=["Date"])

print("Final dataset shape:", final_data.shape)

print("\nColumns:")
print(final_data.columns.tolist())

print("\nWeather missing values after merge:")
print(
    final_data[
        ["Avg_Temperature_C", "Total_Rainfall_mm",
         "Avg_Humidity", "Sensor_Count"]
    ].isna().sum()
)

print("\nDuplicate Date + Mandi_ID + Crop rows:")
print(
    final_data.duplicated(
        subset=["date", "Mandi_ID", "Crop"]
    ).sum()
)

display(final_data.head())
```

```python
duplicates = final_data[
    final_data.duplicated(
        subset=["date", "Mandi_ID", "Crop"],
        keep=False
    )
].sort_values(
    ["date", "Mandi_ID", "Crop"]
)

print("Duplicate rows found:", len(duplicates))
print("Duplicate groups:", duplicates[
    ["date", "Mandi_ID", "Crop"]
].drop_duplicates().shape[0])

display(
    duplicates[
        ["date", "Mandi_ID", "Crop",
         "Modal_Price", "MSP",
         "Avg_Temperature_C", "Total_Rainfall_mm"]
    ].head(20)
)
```

```python
# Remove duplicate Date + Mandi_ID + Crop combinations
final_data = (
    final_data
    .groupby(
        ["date", "Mandi_ID", "Crop"],
        as_index=False
    )
    .agg({
        "Min_Price": "mean",
        "Max_Price": "mean",
        "Modal_Price": "mean",
        "MSP": "mean",
        "Mandi_Name": "first",
        "District": "first",
        "State": "first",
        "Mandi_Type": "first",
        "Total_Area_Acres": "first",
        "Avg_Temperature_C": "first",
        "Total_Rainfall_mm": "first",
        "Avg_Humidity": "first",
        "Sensor_Count": "first"
    })
)

print("Final dataset shape:", final_data.shape)

print("\nDuplicate Date + Mandi_ID + Crop rows:")
print(
    final_data.duplicated(
        subset=["date", "Mandi_ID", "Crop"]
    ).sum()
)

display(final_data.head())
```

```python
print("Missing MSP by crop:")

print(
    final_data.groupby("Crop")["MSP"]
    .apply(lambda x: x.isna().sum())
    .sort_values(ascending=False)
)

print("\nTotal rows by crop:")

print(
    final_data["Crop"]
    .value_counts()
)
```

```python
price_columns = [
    "Min_Price",
    "Max_Price",
    "Modal_Price",
    "MSP"
]

print("Zero or negative values:")

for col in price_columns:
    count = (final_data[col] <= 0).sum()
    print(f"{col}: {count}")
```

```python
# Identify price crash instances
final_data["Price_Crash"] = (
    final_data["Modal_Price"] < final_data["MSP"]
)

print("Total price crash instances:")
print(final_data["Price_Crash"].sum())

print("\nPrice crash rate:")
print(
    round(final_data["Price_Crash"].mean() * 100, 2),
    "%"
)
```

```python
crash_by_crop = (
    final_data.groupby("Crop")["Price_Crash"]
    .agg(
        Crash_Instances="sum",
        Total_Records="count"
    )
)

crash_by_crop["Crash_Rate_%"] = (
    crash_by_crop["Crash_Instances"]
    / crash_by_crop["Total_Records"]
    * 100
)

crash_by_crop = crash_by_crop.sort_values(
    "Crash_Rate_%",
    ascending=False
)

print(crash_by_crop)
```

```python
price_comparison = (
    final_data
    .groupby("Crop")
    .agg(
        Avg_Modal_Price=("Modal_Price", "mean"),
        Avg_MSP=("MSP", "mean")
    )
)

price_comparison["Price_Gap"] = (
    price_comparison["Avg_Modal_Price"]
    - price_comparison["Avg_MSP"]
)

price_comparison["Price_Gap_%"] = (
    price_comparison["Price_Gap"]
    / price_comparison["Avg_MSP"]
    * 100
)

print(price_comparison.round(2))
```

```python
valid_price = master_data[
    master_data["Modal_Price"].notna() &
    master_data["MSP"].notna()
].copy()

crash_by_crop = (
    valid_price
    .groupby("Crop")
    .agg(
        Price_Crash_Instances=("Price_Crash", "sum"),
        Valid_Price_Records=("Price_Crash", "count")
    )
)

crash_by_crop["Crash_Rate_%"] = (
    crash_by_crop["Price_Crash_Instances"]
    / crash_by_crop["Valid_Price_Records"]
    * 100
)

crash_by_crop = crash_by_crop.sort_values(
    "Crash_Rate_%",
    ascending=False
)

print(crash_by_crop.round(2))
```

```python
warehouse_delay = (
    transport
    .dropna(subset=["Transit_Hours"])
    .assign(
        Delayed=lambda x: x["Transit_Hours"] > 18
    )
    .groupby("Destination_Warehouse")
    .agg(
        Total_Trips=("Trip_ID", "count"),
        Delayed_Trips=("Delayed", "sum"),
        Avg_Transit_Hours=("Transit_Hours", "mean")
    )
    .reset_index()
)

warehouse_delay["Delay_Rate_%"] = (
    warehouse_delay["Delayed_Trips"]
    / warehouse_delay["Total_Trips"]
    * 100
)

warehouse_delay = warehouse_delay.sort_values(
    "Delay_Rate_%",
    ascending=False
)

print(warehouse_delay.round(2))
```

```python
mandi_risk = (
    master_data[
        master_data["Modal_Price"].notna() &
        master_data["MSP"].notna()
    ]
    .groupby(["Mandi_ID", "Mandi_Name"])
    .agg(
        Price_Crash_Rate=("Price_Crash", "mean"),
        Avg_Arrivals_Qtl=("Total_Arrivals_Qtl", "mean"),
        Records=("Price_Crash", "count")
    )
    .reset_index()
)

mandi_risk["Price_Crash_Rate_%"] = (
    mandi_risk["Price_Crash_Rate"] * 100
)

mandi_risk = mandi_risk[
    mandi_risk["Records"] >= 10
].sort_values(
    "Price_Crash_Rate_%",
    ascending=False
)

print(
    mandi_risk[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Price_Crash_Rate_%",
            "Avg_Arrivals_Qtl",
            "Records"
        ]
    ].head(15).round(2)
)
```

```python
# Create a simple 0–100 risk score for each mandi

mandi_risk["Price_Risk_Score"] = (
    mandi_risk["Price_Crash_Rate_%"]
    / mandi_risk["Price_Crash_Rate_%"].max()
    * 100
)

mandi_risk["Supply_Score"] = (
    mandi_risk["Avg_Arrivals_Qtl"]
    / mandi_risk["Avg_Arrivals_Qtl"].max()
    * 100
)

# Higher supply + higher price risk = greater intervention priority
mandi_risk["Priority_Score"] = (
    0.6 * mandi_risk["Price_Risk_Score"] +
    0.4 * mandi_risk["Supply_Score"]
)

mandi_risk = mandi_risk.sort_values(
    "Priority_Score",
    ascending=False
)

print(
    mandi_risk[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Price_Crash_Rate_%",
            "Avg_Arrivals_Qtl",
            "Priority_Score"
        ]
    ].head(15).round(2)
)
```

```python
# Calculate average transit time for each mandi
transport_mandi_risk = (
    transport
    .dropna(subset=["Transit_Hours"])
    .groupby("Mandi_ID")
    .agg(
        Avg_Transit_Hours=("Transit_Hours", "mean"),
        Total_Trips=("Trip_ID", "count"),
        Delayed_Trips=("Transit_Hours", lambda x: (x > 18).sum())
    )
    .reset_index()
)

transport_mandi_risk["Delay_Rate_%"] = (
    transport_mandi_risk["Delayed_Trips"]
    / transport_mandi_risk["Total_Trips"]
    * 100
)

# Add transport information to our mandi risk table
mandi_risk_final = mandi_risk.merge(
    transport_mandi_risk,
    on="Mandi_ID",
    how="left"
)

print(
    mandi_risk_final[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Price_Crash_Rate_%",
            "Avg_Arrivals_Qtl",
            "Avg_Transit_Hours",
            "Delay_Rate_%"
        ]
    ].sort_values(
        "Delay_Rate_%",
        ascending=False
    ).head(15).round(2)
)
```

```python
# Normalize the three risk factors to 0–100

mandi_risk_final["Price_Risk_Score"] = (
    mandi_risk_final["Price_Crash_Rate_%"]
    / mandi_risk_final["Price_Crash_Rate_%"].max()
    * 100
)

mandi_risk_final["Supply_Pressure_Score"] = (
    mandi_risk_final["Avg_Arrivals_Qtl"]
    / mandi_risk_final["Avg_Arrivals_Qtl"].max()
    * 100
)

mandi_risk_final["Transport_Risk_Score"] = (
    mandi_risk_final["Delay_Rate_%"]
    / mandi_risk_final["Delay_Rate_%"].max()
    * 100
)

# Final optimizer priority score
mandi_risk_final["Priority_Score"] = (
    0.50 * mandi_risk_final["Price_Risk_Score"] +
    0.30 * mandi_risk_final["Supply_Pressure_Score"] +
    0.20 * mandi_risk_final["Transport_Risk_Score"]
)

final_priority = (
    mandi_risk_final
    .sort_values("Priority_Score", ascending=False)
)

print(
    final_priority[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Price_Crash_Rate_%",
            "Avg_Arrivals_Qtl",
            "Delay_Rate_%",
            "Priority_Score"
        ]
    ].head(15).round(2)
)
```

```python
# Select the final optimizer scores
optimizer_scores = mandi_risk_final[
    [
        "Mandi_ID",
        "Price_Risk_Score",
        "Supply_Pressure_Score",
        "Transport_Risk_Score",
        "Priority_Score"
    ]
].drop_duplicates("Mandi_ID")

# Add optimizer scores to the integrated dataset
powerbi_data = master_data.merge(
    optimizer_scores,
    on="Mandi_ID",
    how="left"
)

print("Power BI dataset shape:", powerbi_data.shape)
print("\nColumns:")
print(powerbi_data.columns.tolist())

print("\nPreview:")
print(powerbi_data.head())
```

```python
# Rank mandis from highest to lowest Priority Score
mandi_risk_final["Priority_Rank"] = (
    mandi_risk_final["Priority_Score"]
    .rank(method="first", ascending=False)
    .astype(int)
)

# Assign priority based on rank
def assign_priority_rank(rank):
    if rank <= 10:
        return "Critical"
    elif rank <= 25:
        return "High"
    elif rank <= 42:
        return "Medium"
    else:
        return "Low"

mandi_risk_final["Priority_Level"] = (
    mandi_risk_final["Priority_Rank"]
    .apply(assign_priority_rank)
)

print(
    mandi_risk_final[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Priority_Score",
            "Priority_Rank",
            "Priority_Level"
        ]
    ]
    .sort_values("Priority_Rank")
    .head(20)
)
```

```python
def generate_reason(row):
    reasons = []

    if row["Price_Risk_Score"] >= 70:
        reasons.append("High price risk")

    if row["Supply_Pressure_Score"] >= 70:
        reasons.append("High supply pressure")

    if row["Transport_Risk_Score"] >= 70:
        reasons.append("High transport risk")

    if not reasons:
        reasons.append("Moderate combined risk")

    return " + ".join(reasons)


mandi_risk_final["Recommendation_Reason"] = (
    mandi_risk_final.apply(generate_reason, axis=1)
)

print(
    mandi_risk_final[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Priority_Level",
            "Priority_Score",
            "Recommendation_Reason"
        ]
    ]
    .sort_values("Priority_Score", ascending=False)
    .head(15)
)
```

```python
def generate_action(row):
    actions = []

    if row["Price_Risk_Score"] >= 70:
        actions.append("Monitor price and MSP risk")

    if row["Supply_Pressure_Score"] >= 70:
        actions.append("Prioritize supply movement")

    if row["Transport_Risk_Score"] >= 70:
        actions.append("Prioritize transport")

    if not actions:
        actions.append("Continue routine monitoring")

    return " + ".join(actions)


mandi_risk_final["Recommended_Action"] = (
    mandi_risk_final.apply(generate_action, axis=1)
)

print(
    mandi_risk_final[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Priority_Level",
            "Priority_Score",
            "Recommendation_Reason",
            "Recommended_Action"
        ]
    ]
    .sort_values("Priority_Score", ascending=False)
    .head(15)
)
```
