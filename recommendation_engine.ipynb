```python
import pandas as pd

optimizer = pd.read_csv("mandi_optimizer_results.csv")

print("Optimizer loaded successfully!")
print("Shape:", optimizer.shape)
print(optimizer.head())
```

```python
def get_top_mandis(n=5):
    return optimizer.sort_values(
        "Priority_Score",
        ascending=False
    ).head(n)[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Priority_Level",
            "Priority_Score",
            "Recommendation_Reason",
            "Recommended_Action"
        ]
    ]


print(get_top_mandis(5).to_string(index=False))
```

```python
def get_mandis_by_priority(level):
    result = optimizer[
        optimizer["Priority_Level"].str.lower() == level.lower()
    ].sort_values(
        "Priority_Score",
        ascending=False
    )

    return result[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Priority_Level",
            "Priority_Score",
            "Recommendation_Reason",
            "Recommended_Action"
        ]
    ]


print(get_mandis_by_priority("Critical").to_string(index=False))
```

```python
def find_mandi(mandi_name):
    result = optimizer[
        optimizer["Mandi_Name"].str.contains(
            mandi_name,
            case=False,
            na=False
        )
    ]

    return result[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Priority_Level",
            "Priority_Score",
            "Recommendation_Reason",
            "Recommended_Action"
        ]
    ].sort_values(
        "Priority_Score",
        ascending=False
    )


print(find_mandi("Orai").to_string(index=False))
```

```python
def get_highest_transport_risk(n=5):
    return optimizer.sort_values(
        "Transport_Risk_Score",
        ascending=False
    ).head(n)[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Transport_Risk_Score",
            "Priority_Level",
            "Priority_Score",
            "Recommended_Action"
        ]
    ]


print(get_highest_transport_risk(5).to_string(index=False))
```

```python
def get_highest_price_risk(n=5):
    return optimizer.sort_values(
        "Price_Risk_Score",
        ascending=False
    ).head(n)[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Price_Risk_Score",
            "Priority_Level",
            "Priority_Score",
            "Recommended_Action"
        ]
    ]


print(get_highest_price_risk(5).to_string(index=False))
```

```python
def get_highest_supply_pressure(n=5):
    return optimizer.sort_values(
        "Supply_Pressure_Score",
        ascending=False
    ).head(n)[
        [
            "Mandi_ID",
            "Mandi_Name",
            "Supply_Pressure_Score",
            "Priority_Level",
            "Priority_Score",
            "Recommended_Action"
        ]
    ]


print(get_highest_supply_pressure(5).to_string(index=False))
```

```python
def recommend(query="top"):
    
    query = query.lower().strip()

    if "transport" in query:
        result = get_highest_transport_risk(5)

    elif "price" in query or "msp" in query:
        result = get_highest_price_risk(5)

    elif "supply" in query or "arrival" in query:
        result = get_highest_supply_pressure(5)

    elif "critical" in query:
        result = get_mandis_by_priority("Critical")

    elif "high" in query:
        result = get_mandis_by_priority("High")

    else:
        result = get_top_mandis(5)

    return result


print(
    recommend("Which mandis have the highest transport risk?")
    .to_string(index=False)
)
```

```python
def recommend(query="top"):
    
    query = query.lower().strip()

    # Check for specific mandi name
    mandi_matches = optimizer[
        optimizer["Mandi_Name"].str.lower().str.contains(
            query,
            na=False
        )
    ]

    if len(mandi_matches) > 0:
        return mandi_matches[
            [
                "Mandi_ID",
                "Mandi_Name",
                "Priority_Level",
                "Priority_Score",
                "Recommendation_Reason",
                "Recommended_Action"
            ]
        ]

    # Risk-based queries
    if "transport" in query:
        return get_highest_transport_risk(5)

    elif "price" in query or "msp" in query:
        return get_highest_price_risk(5)

    elif "supply" in query or "arrival" in query:
        return get_highest_supply_pressure(5)

    elif "critical" in query:
        return get_mandis_by_priority("Critical")

    elif "high" in query:
        return get_mandis_by_priority("High")

    else:
        return get_top_mandis(5)


print(
    recommend("Orai").to_string(index=False)
)
```

```python
def explain_mandi(mandi_name):
    result = optimizer[
        optimizer["Mandi_Name"].str.contains(
            mandi_name,
            case=False,
            na=False
        )
    ]

    if result.empty:
        return f"No mandi found matching '{mandi_name}'."

    row = result.sort_values(
        "Priority_Score",
        ascending=False
    ).iloc[0]

    print(f"\n{row['Mandi_Name']} — {row['Priority_Level']}")
    print(f"Priority Score: {row['Priority_Score']:.2f}")
    print(f"Main Risks: {row['Recommendation_Reason']}")
    print(f"Recommended Action: {row['Recommended_Action']}")


explain_mandi("Orai")
```

```python
def ai_recommendation(query):
    query = query.lower().strip()

    # Specific mandi
    mandi_match = optimizer[
        optimizer["Mandi_Name"].str.lower().str.contains(
            query,
            na=False
        )
    ]

    if not mandi_match.empty:
        row = mandi_match.sort_values(
            "Priority_Score",
            ascending=False
        ).iloc[0]

        return (
            f"{row['Mandi_Name']} — {row['Priority_Level']}\n"
            f"Priority Score: {row['Priority_Score']:.2f}\n"
            f"Main Risks: {row['Recommendation_Reason']}\n"
            f"Recommended Action: {row['Recommended_Action']}"
        )

    # General queries
    if "transport" in query:
        result = get_highest_transport_risk(5)

    elif "price" in query or "msp" in query:
        result = get_highest_price_risk(5)

    elif "supply" in query or "arrival" in query:
        result = get_highest_supply_pressure(5)

    elif "critical" in query:
        result = get_mandis_by_priority("Critical")

    else:
        result = get_top_mandis(5)

    return result


print(ai_recommendation("Orai"))
```

```python
def generate_mandi_response(mandi_name):
    result = optimizer[
        optimizer["Mandi_Name"].str.contains(
            mandi_name,
            case=False,
            na=False
        )
    ]

    if result.empty:
        return f"No mandi found matching '{mandi_name}'."

    row = result.sort_values(
        "Priority_Score",
        ascending=False
    ).iloc[0]

    response = {
        "Mandi": row["Mandi_Name"],
        "Priority": row["Priority_Level"],
        "Priority_Score": round(row["Priority_Score"], 2),
        "Price_Risk": round(row["Price_Risk_Score"], 2),
        "Supply_Risk": round(row["Supply_Pressure_Score"], 2),
        "Transport_Risk": round(row["Transport_Risk_Score"], 2),
        "Reason": row["Recommendation_Reason"],
        "Action": row["Recommended_Action"]
    }

    return response


print(generate_mandi_response("Orai"))
```

```python
def recommend(query="top"):
    query = query.lower().strip()

    # Specific mandi name
    mandi_matches = optimizer[
        optimizer["Mandi_Name"].str.lower().str.contains(
            query,
            na=False
        )
    ]

    if not mandi_matches.empty:
        row = mandi_matches.sort_values(
            "Priority_Score",
            ascending=False
        ).iloc[0]

        return (
            f"{row['Mandi_Name']} — {row['Priority_Level']}\n"
            f"Priority Score: {row['Priority_Score']:.2f}\n"
            f"Main Risks: {row['Recommendation_Reason']}\n"
            f"Recommended Action: {row['Recommended_Action']}"
        )

    # Transport-related questions
    if any(word in query for word in [
        "transport", "delay", "delivery", "logistics"
    ]):
        return get_highest_transport_risk(5)

    # Price/MSP-related questions
    elif any(word in query for word in [
        "price", "msp", "crash", "market price"
    ]):
        return get_highest_price_risk(5)

    # Supply/arrival-related questions
    elif any(word in query for word in [
        "supply", "arrival", "arrivals", "quantity"
    ]):
        return get_highest_supply_pressure(5)

    # Critical/urgent questions
    elif any(word in query for word in [
        "critical", "urgent", "immediate", "attention", "risk"
    ]):
        return get_top_mandis(5)

    # Default
    else:
        return get_top_mandis(5)
```
