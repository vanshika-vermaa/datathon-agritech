```python
import pandas as pd

# Load both files
df_arrivals = pd.read_csv("cleaned_mandi_arrivals.csv")
df_master   = pd.read_csv("cleaned_mandi_master.csv")

# Merge on mandi_id
df_joined = pd.merge(
    df_arrivals,
    df_master,
    on="Mandi_ID",
    how="left"   # keep all arrivals, add master info
)

# Sort if needed
df_joined = df_joined.sort_values(["Date", "Mandi_ID"]).reset_index(drop=True)

# Save to a new file (different from cleaned arrivals)
df_joined.to_csv("joined_mandi_data.csv", index=False)

print("Joined dataset saved as 'joined_mandi_data.csv'")
```
