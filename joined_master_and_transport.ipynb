```python
import pandas as pd

# Load both files
df_transport = pd.read_csv("cleaned_transport_logistics.csv")
df_master    = pd.read_csv("cleaned_mandi_master.csv")

# Merge on mandi_id
df_transport_joined = pd.merge(
    df_transport,
    df_master,
    on="Mandi_ID",
    how="left"   # keep all transport records, add master info
)

# Sort if needed
df_transport_joined = df_transport_joined.sort_values(["Mandi_ID"]).reset_index(drop=True)

# Save to a new file
df_transport_joined.to_csv("transport_with_master.csv", index=False)

print("Joined dataset saved as 'transport_with_master.csv'")
```

```python
# Check joined dataset

print("Transport + Mandi Master")
print("Shape:", df_transport_joined.shape)

print("\nColumns:")
print(df_transport_joined.columns.tolist())

print("\nMissing values after join:")
print(df_transport_joined.isna().sum())

print("\nSample data:")
display(df_transport_joined.head())
```
